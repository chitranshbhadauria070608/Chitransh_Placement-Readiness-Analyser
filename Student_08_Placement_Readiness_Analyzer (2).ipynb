```python
# TODO: Display the shape and column names
print("dataset shape (row,column):", df.shape)
print("column names:" , list(df.columns))


# TODO: Inspect data types and missing values
df.info() 
print("-----------------------------------")
print("missing value in each columns")
print(df.isnull().sum())
# TODO: Count duplicate rows
print("number of duplicate rows:",df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum
sections = ["Aptitude", "Reasoning", "Verbal", "Technical"]
marks = df[sections].to_numpy()

print("Overall average:", np.mean(marks))
print("Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))



```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s) 
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
# Create Calculated Columns

df["Total_Score"] = (
    df["Aptitude"] +
    df["Reasoning"] +
    df["Verbal"] +
    df["Technical"]
)

df["Average_Score"] = df["Total_Score"] / 4

# Useful Summary
print("Average Total Score:", df["Total_Score"].mean())
print("Highest Total Score:", df["Total_Score"].max())

# Filter Records
high_performers = df[df["Average_Score"] >= 80]

print("\nStudents with Average Score >= 80")
display(high_performers)

# Sort Records
print("\nTop 5 Students")
display(
    df.sort_values(by="Average_Score", ascending=False)
      [["Name", "Average_Score"]]
      .head()
)

print("\nBottom 5 Students")
display(
    df.sort_values(by="Average_Score", ascending=True)
      [["Name", "Average_Score"]]
      .head()
)

```
